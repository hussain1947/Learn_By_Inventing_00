```python
def finda(equation):
  a,b=equation
  if a==0:
    print('Division by Zero')
    return None
  return b/a
```

```python
finda([2,8])
```

```python
def findxy(equations):
  e1=equations[0]
  e2=equations[1]
  find_multiplier=e1[0]/e2[0]
  equation_y=[ (e1[1]) - (e2[1]*find_multiplier) , (e1[2]) - (e2[2]*find_multiplier) ]
  y= finda(equation_y)
  equation_x=[ e1[0], e1[2]-y*e1[1]]
  x= finda(equation_x)
  return x,y

```

```python
equations =[[2,1,10],[1,-1,1]]
```

```python
findxy(equations)
```

```python
def swap(a,b):
  temp=a
  a=b
  b=temp
  return a,b
```
