```python
base,number,expo = 10,100,0
low,high=1,2

def find_power(low,high):
  mid=(low+high)/2
  print(f'Now the  low:{low},             High:{high},              Mid is {mid}            and base**mid is {base**mid}')

  if 10**mid == number:
    expo = mid

  elif base**mid > number:
    print(base**mid)
    high = mid
    return find_power(low,high)

  elif base**mid <number:
    low = mid
    return find_power(low,high)

  return expo

x=find_power(low,high)
print(f'Therefore the Log value for {number} when base={base} is: {x} ')
```
