```python
### This Funcitons Returns Average without Overflow Error ###
### Minimum Requires Two Input parameters to process Avg  ###
def my_avg(x1, x2):
    return x1 + ((x2 - x1) / 2)
```

```python
### This Funcitons Returns Square root value             ###
### Minimum Requires One Input parameter to process Avg  ###
def my_sqrt(num):
    low, high = 1, num
    guess = my_avg(low, high)
    for i in range(20):
        if guess * guess < num:
            low = guess
        else:
            high = guess
        guess = my_avg(low, high)
    return guess
```

```python
### This Funcitons Returns Cube root value ###
### Minimum Requires One Input parameter to process Avg  ###
def my_cubert(num):
  low,high=1,num
  guess = my_avg(low,high)
  for i in range(20):
    if guess*guess*guess < num:
        low= guess
    else:
        high= guess
    guess = my_avg(low,high)
  return guess
```

```python
def my_hypotenuse_2d(x1,y1,x2,y2):
  base=x2-x1
  height= y2-y1
  print(f'The base is {base} and height is {height}')
  hypotenuse = my_sqrt(base*base+height *height)
  print(f'The hypotenuse is {hypotenuse}')
  return hypotenuse
```

```python
my_hypotenuse_2d(-2,2,1,6)
```

```python
def my_hypotenuse_3d(x1,y1,z1,x2,y2,z2):

  base=x2-x1
  depth= z2-z1
  print(f'For First  Traingle The base is {base} and depth is {depth}')
  hypotenuse_1 = my_sqrt(base*base+depth *depth)
  print(f'For First  Traingle The hypotenuse is {hypotenuse_1}')

  new_base=hypotenuse_1
  height= y2-y1
  print(f'For Second Traingle The new_base is {new_base} and height is {height}')
  hypotenuse_2 = my_sqrt(new_base*new_base+height*height)
  print(f'For Second Traingle The hypotenuse is {hypotenuse_2}')
  return hypotenuse_2
```

```python
my_hypotenuse_3d(1,1,1,3,3,3)
```

Imagine a taxi that can only drive along a grid of streets — no diagonal shortcuts. The distance it travels between two points is called the Manhattan distance.

```python
def manhattan_distance(x1,y1,x2,y2):
  return abs(x2-x1)+abs(y2-y1)
```

```python

print(my_hypotenuse_2d(2, 3, 5, 1))
print(manhattan_distance(2, 3, 5, 1))



print(my_hypotenuse_2d(0, 0, 0, 0))
print(manhattan_distance(0, 0, 0, 0))


print(my_hypotenuse_2d(-1, -1, 1, 1))
print(manhattan_distance(-1, -1, 1, 1))
```
