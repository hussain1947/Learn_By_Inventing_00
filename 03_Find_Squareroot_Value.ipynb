```python
number = 49
low = 0
high = number

def find_square_root(low, high):
    mid = (low + high) / 2

    print(f"Now the Mid is {mid} and mid**2 is {mid**2}")

    if mid**2 == number:
        return mid

    elif mid**2 > number:
        high = mid
        return find_square_root(low, high)

    elif mid**2 < number:
        low = mid
        return find_square_root(low, high)

x = find_square_root(low, high)

print(f"Square root is {x}")
```
